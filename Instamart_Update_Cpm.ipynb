```python
import time
import random
import re
import json
import imaplib
import email
import datetime
import pandas as pd
from bs4 import BeautifulSoup

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


# ==============================
# 🔧 HUMAN TYPING
# ==============================
def human_typing(element, text):
    element.clear()
    for char in text:
        element.send_keys(char)
        time.sleep(random.uniform(0.05, 0.15))


# ==============================
# 📩 OTP EXTRACTOR
# ==============================
def extract_instamart_otp(sender, subject_keyword, email_user, email_pass):
    try:
        mail = imaplib.IMAP4_SSL("imap.gmail.com")
        mail.login(email_user, email_pass)
        mail.select("INBOX")

        since_date = (datetime.datetime.now() - datetime.timedelta(days=1)).strftime("%d-%b-%Y")

        status, messages = mail.search(
            None,
            f'(FROM "{sender}" SUBJECT "{subject_keyword}" SINCE "{since_date}")'
        )

        email_ids = messages[0].split()
        if not email_ids:
            raise Exception("No OTP email found")

        latest_email_id = email_ids[-1]
        status, msg_data = mail.fetch(latest_email_id, "(RFC822)")
        msg = email.message_from_bytes(msg_data[0][1])

        otp = None

        for part in msg.walk():
            content_type = part.get_content_type()

            if content_type in ["text/plain", "text/html"]:
                body = part.get_payload(decode=True)
                if body:
                    text = body.decode(errors="ignore")

                    if content_type == "text/html":
                        soup = BeautifulSoup(text, "html.parser")
                        text = soup.get_text()

                    match = re.search(r"\b\d{4,6}\b", text)
                    if match:
                        otp = match.group()
                        break

        mail.logout()

        if not otp:
            raise Exception("OTP not found in email")

        return otp

    except Exception as e:
        print("❌ OTP Error:", e)
        return None


# ==============================
# 🔐 GET AUTH TOKEN
# ==============================
def get_auth_token(driver):
    logs = driver.get_log("performance")

    for entry in logs:
        log = json.loads(entry["message"])["message"]

        if log["method"] == "Network.requestWillBeSent":
            headers = log.get("params", {}).get("request", {}).get("headers", {})

            auth = headers.get("authorization") or headers.get("Authorization")
            if auth:
                return auth

    return None


# ==============================
# 🚀 MAIN LOOP
# ==============================
for index, row in df_brands.iterrows():

    # brand = row['brand']
    # email_1 = row['email']
    # sender = row['Sender']

    brand      = row['brand']
    email_1    = row['email']
    ad_account = row['ad_account_id']
    sender     = row['Sender']
        

    print(f"\n🚀 Processing: {brand}")

    authorization = None   # ✅ critical fix

    # ==============================
    # 🌐 SELENIUM SETUP
    # ==============================
    options = Options()
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")

    options.set_capability("goog:loggingPrefs", {"performance": "ALL"})

    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Network.enable", {})

    wait = WebDriverWait(driver, 25)

    try:
        # ==============================
        # 🔐 LOGIN FLOW (FIXED)
        # ==============================
        driver.get("https://partner.instamart.in/login")
        time.sleep(5)

        # Enter email FIRST
        email_input = wait.until(
            EC.visibility_of_element_located((By.ID, "email-input"))
        )
        human_typing(email_input, email_1)

        # Click Send OTP
        send_otp_btn = wait.until(
            EC.element_to_be_clickable((By.XPATH, "//button[.//span[text()='Send OTP']]"))
        )
        send_otp_btn.click()

        print("📩 OTP requested...")
        time.sleep(15)

        # ==============================
        # 🔢 FETCH OTP
        # ==============================
        otp = extract_instamart_otp(
            sender,
            "Your Login OTP for Swiggy Instamart Ads Portal",
            Alert_email,
            Alert_password
        )

        if not otp:
            raise Exception("OTP not received")

        print("✅ OTP:", otp)

        # Enter OTP
        for i, digit in enumerate(otp, start=1):
            box = wait.until(
                EC.visibility_of_element_located((By.NAME, f"otp{i}"))
            )
            box.send_keys(digit)

        # Click login
        login_btn = wait.until(
            EC.element_to_be_clickable((By.XPATH, "//span[text()='Login']/ancestor::button"))
        )
        login_btn.click()

        print("🔓 Logged in...")
        time.sleep(5)

        # ==============================
        # 🧠 SELECT BRAND
        # ==============================
        brand_el = wait.until(
            EC.element_to_be_clickable((By.XPATH, f"//div[normalize-space()='{brand}']"))
        )
        brand_el.click()

        time.sleep(3)

        # ==============================
        # 🔁 TRIGGER API FOR TOKEN
        # ==============================
        driver.get("https://partner.instamart.in/instamart/advertisement")
        time.sleep(5)

        authorization = get_auth_token(driver)

        if not authorization:
            raise Exception("Authorization token not captured")

        print("✅ AUTH TOKEN:", authorization[:50], "...")

    except Exception as e:
        print("❌ ERROR:", e)
        driver.quit()
        continue

    finally:
        driver.quit()

    # ==============================
    # 🧪 SAFE CHECK BEFORE API
    # ==============================
    if not authorization:
        print("❌ Skipping API — No auth token")
        continue
        
    ## FETCH MIN_BID AND SEARCH_COUNT
    import psycopg2
    import requests
    import uuid
    import time
    # import datetime
    import yagmail

    # ==============================
    # 🔐 CONFIG
    # ==============================

    import requests
    import json

    url = "https://brand-portal-service-http.swiggy.com/api/v1/suggest/keyword/bids"

    headers = {
        "accept": "*/*",
        "accept-language": "en-US,en;q=0.9",
        "app_version": "1.4.103",
        "authorization": authorization,  
        "cache-control": "no-cache",
        "content-type": "application/json",
        "origin": "https://partner.instamart.in",
        "pragma": "no-cache",
        "referer": "https://partner.instamart.in/",
        "user-agent": "Mozilla/5.0",
        "x-client-account-id": ad_account,
        "x-client-id": "IM_ADS_EXTERNAL_DASHBOARD",
        "x-client-request-id": "a38bac14-0142-47a1-aec9-0e409a073f8a",
        "x-signature": "06a02743d2af4db755c18929e0daca7b7ba67425d14f1de74a1fdba767d97061",
        "x-timestamp": "1776062335442"
    }
    # ==============================
    # 🛠 DB CONNECTION
    # ==============================

    try:
        print("Connecting with Voylla DB at:",
              datetime.datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S'))

        conn = psycopg2.connect(**Voylla_config)
        cursor = conn.cursor()

        print("Voylla DB Connected at:",
              datetime.datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S'))

    except psycopg2.DatabaseError as error:
        body = f'Error connecting to db: {str(error)}'
        print(body)

    # ==============================
    # 📥 FETCH CAMPAIGNS
    # ==============================

    try:
        cursor.execute("""
            SELECT DISTINCT
                "CAMPAIGN_ID",
                "CAMPAIGN_NAME",
                "KEYWORD"
            FROM voylla."Instamart_Campaign_Placement"
            WHERE "Date"::DATE >= CURRENT_DATE - INTERVAL '7 day'
        """)

        rows = cursor.fetchall()

    except Exception as e:
        print("Error fetching campaigns:", e)
        raise SystemExit()

    # ==============================
    # 🧠 GROUP DATA
    # ==============================

    campaigns = {}

    for campaign_id, campaign_name, keyword in rows:
        if campaign_id not in campaigns:
            campaigns[campaign_id] = {
                "campaign_name": campaign_name,
                "keywords": set()
            }

        campaigns[campaign_id]["keywords"].add(keyword)

    # ==============================
    # 🔧 PAYLOAD BUILDER
    # ==============================

    def create_payload(keywords):
        return {
            "account_id": ad_account,
            "ad_unit_type": "PRODUCT",
            "bidding_strategy": {"cpm": {}},
            "candidate_ids": [
                "ZSODFATYY0", "5MBCONG6OL", "WF62D7VJVH", "HD0K3Q9SNV",
                "PJ4Y59XNG7", "1RYOMNJN42", "8ADJ0O3PJZ", "RDA2979R2H", "40XA7DSMRC"
            ],
            "days_of_week": [
                "MONDAY", "TUESDAY", "WEDNESDAY", "THURSDAY",
                "FRIDAY", "SATURDAY", "SUNDAY"
            ],
            "end_time": None,
            "keyword_criteria": {
                "keyword_criteria": [
                    {"keyword": kw, "match_type": mt}
                    for kw in keywords
                    for mt in ["KEYWORD_MATCH_TYPE_BROAD", "KEYWORD_MATCH_TYPE_EXACT"]
                ]
            },
            "location_info": [],
            "meal_slots": [
                "MEAL_SLOT_EARLY_BREAKFAST","MEAL_SLOT_BREAKFAST",
                "MEAL_SLOT_LUNCH","MEAL_SLOT_SNACKS",
                "MEAL_SLOT_DINNER","MEAL_SLOT_LATE_DINNER"
            ],
            "placement_ids": ["8bff0282-f639-4aa1-94b8-5531f34c0c55"],
            "request_context": {
                "client_id": "IM_ADS_EXTERNAL_DASHBOARD",
                "request_id": str(uuid.uuid4()),
                "request_time": datetime.datetime.utcnow().isoformat() + "Z"
            },
            "start_time": "2026-02-05T18:30:00.000Z"
        }

    # ==============================
    # 🚀 MAIN PROCESS
    # ==============================

    for campaign_id, data in campaigns.items():
        try:
            print(f"\nProcessing Campaign: {data['campaign_name']}")

            payload = create_payload(list(data["keywords"]))

            response = requests.post(url, headers=headers, json=payload)

            if response.status_code != 200:
                raise Exception(f"API Error {response.status_code}: {response.text}")

            res_json = response.json()


            for item in res_json.get("bidResponses", []):

                keyword = item["keywordCriteria"]["keyword"]
                match_type = item["keywordCriteria"]["matchType"]

                nanos = item["bidSuggestion"]["minimumBidAllowed"]["nanos"]

                # 🔥 Correct conversion
                min_bid = nanos / 1_000_000

                search_count = item.get("searchCount", 0)

                print(f"{keyword} | {match_type} | Min Bid: {min_bid} | Searches: {search_count}")
                cursor.execute("""
                    INSERT INTO voylla.instamart_bid 
                (campaign_id, campaign_name, targeting, match_type, min_bid, search_count, "Brand")
                VALUES (%s, %s, %s, %s, %s, %s, %s)
                ON CONFLICT (campaign_id, targeting, match_type,"Brand")
                DO UPDATE SET 
                    min_bid = EXCLUDED.min_bid,
                    search_count = EXCLUDED.search_count;
            """, (
                campaign_id,
                data["campaign_name"],
                keyword,
                match_type,
                min_bid,
                search_count,
                brand ))

            conn.commit()
            print("✅ Success")

        except Exception as e:
            error_msg = f"Error in campaign {campaign_id}: {str(e)}"
            print(error_msg)

            # 📧 send alert
            with yagmail.SMTP(Alert_email, Alert_password) as yag:
                subject = SD['Script_Name'] + " - Processing Error"
                yag.send(
                    to=receiver_email,
                    subject=subject,
                    contents=error_msg,
                    headers={"From": f"{Alert_sender_name} <{Alert_email}>"}
                )

    # ==============================
    # 🔚 CLEANUP
    # ==============================

    cursor.close()
    conn.close()

    print("\n🎉 All campaigns processed successfully!")
    
    ## FETCH CURRENT_BID
    
    import requests
    import json

    url = "https://brand-portal-service-http.swiggy.com/api/v1/campaigns"

    headers = {
        "accept": "/",
        "content-type": "application/json",
        "authorization": authorization, 
        "origin": "https://partner.instamart.in",
        "referer": "https://partner.instamart.in/",
        "user-agent": "Mozilla/5.0",
        "x-client-account-id": ad_account,
        "x-client-id": "IM_ADS_EXTERNAL_DASHBOARD",
        "x-client-request-id": "740c074b-3404-4aeb-bb29-65dfc9404552",
        "x-signature": "1d9ccbca35a03cc260ccc065dbf59a55b498bd88e2f154ef2becb737f06f4946",
        "x-timestamp": "1776064495350"
    }

    payload = {
        "request_context": {
            "request_id": "c356eaa0-a204-497a-aa73-c457b9f7a318",
            "client_id": "IM_ADS_EXTERNAL_DASHBOARD"
        },
        "account_id": ad_account,
        "metrics": [
            {"name": "METRIC_TYPE_IMPRESSIONS"},
            {"name": "METRIC_TYPE_CLICKS"},
            {"name": "METRIC_TYPE_CTR"},
            {"name": "METRIC_TYPE_ADD_TO_CART_COUNT"},
            {"name": "METRIC_TYPE_CONVERSIONS"},
            {"name": "METRIC_TYPE_CONVERSION_RATE"},
            {"name": "METRIC_TYPE_GMV"},
            {"name": "METRIC_TYPE_ROI"},
            {"name": "METRIC_TYPE_BUDGET_BURNT_REALTIME"},
            {"name": "METRIC_TYPE_DAILY_BUDGET_BURNT_REALTIME"},
            {"name": "METRIC_TYPE_LAST_ACTIVE_DAY_UPTIME_EFFICIENCY"}
        ],
        "filters": {
            "logical_operator": "LOGICAL_OPERATOR_AND",
            "filters": [
                {
                    "campaign_search_filter": "CAMPAIGN_SEARCH_FILTER_TYPE_ACCOUNT_ID",
                    "comparison_operator": "COMPARISON_OPERATOR_EQUAL",
                    "values": [{"s_value": "6da7a6d7-9265-4d70-9ced-e5cc6bc8309b"}]
                },
                {
                    "campaign_search_filter": "CAMPAIGN_SEARCH_FILTER_TYPE_START_TIME",
                    "comparison_operator": "COMPARISON_OPERATOR_LESS_THAN_OR_EQUAL",
                    "values": [{"i_value": 1776018599}]
                },
                {
                    "campaign_search_filter": "CAMPAIGN_SEARCH_FILTER_TYPE_END_TIME",
                    "comparison_operator": "COMPARISON_OPERATOR_GREATER_THAN_OR_EQUAL",
                    "values": [{"i_value": 1774981800}]
                },
                {
                    "campaign_search_filter": "CAMPAIGN_SEARCH_FILTER_TYPE_CAMPAIGN_TYPE",
                    "comparison_operator": "COMPARISON_OPERATOR_EQUAL",
                    "values": [
                        {"s_value": "CAMPAIGN_TYPE_ITEM"},
                        {"s_value": "CAMPAIGN_TYPE_BANNER"},
                        {"s_value": "CAMPAIGN_TYPE_SEARCH_AUTO_SUGGEST"},
                        {"s_value": "CAMPAIGN_TYPE_COLLECTION_ADS"}
                    ]
                },
                {
                    "campaign_search_filter": "CAMPAIGN_SEARCH_FILTER_TYPE_PLACEMENT",
                    "comparison_operator": "COMPARISON_OPERATOR_EQUAL",
                    "values": [
                        {"s_value": "KBA"}, {"s_value": "BBA"}, {"s_value": "YGTI"},
                        {"s_value": "LMAO"}, {"s_value": "FBT"}, {"s_value": "BRANDS_IN_FOCUS"},
                        {"s_value": "HOUSEHOLD_FAVORITES"}, {"s_value": "IM_SPECIAL_OCCASION"},
                        {"s_value": "MUST_TRY"}, {"s_value": "ON_FOCUS_TODAY"},
                        {"s_value": "L2_BANNER"}, {"s_value": "SILB"}, {"s_value": "BOTD"},
                        {"s_value": "TOP_SLOT"}, {"s_value": "HOME_PAGE_COLLECTIONS"},
                        {"s_value": "CART_PAGE_COLLECTIONS"}
                    ]
                },
                {
                    "campaign_search_filter": "CAMPAIGN_SEARCH_FILTER_TYPE_CAMPAIGN_STATUS",
                    "comparison_operator": "COMPARISON_OPERATOR_EQUAL",
                    "values": [
                        {"s_value": "CAMPAIGN_STATUS_ACTIVE"},
                        {"s_value": "CAMPAIGN_STATUS_IN_REVIEW"},
                        {"s_value": "CAMPAIGN_STATUS_UPCOMING"},
                        {"s_value": "CAMPAIGN_STATUS_REJECTED"},
                        {"s_value": "CAMPAIGN_STATUS_STOPPED"},
                        {"s_value": "CAMPAIGN_STATUS_EXHAUSTED"},
                        {"s_value": "CAMPAIGN_STATUS_EXPIRED"},
                        {"s_value": "CAMPAIGN_STATUS_PAUSED"}
                    ]
                }
            ]
        },
        "pagination_context": {"offset": "0", "size": 30},
        "sort": {
            "sort_order": "SORT_ORDER_DESC",
            "ads_campaign_attribute": "ADS_CAMPAIGN_ATTRIBUTE_END_DATE"
        }
    }

    response = requests.post(url, headers=headers, json=payload)
    data = response.json()
    
    def extract_keyword_bids(data):
        results = []
        campaigns = data.get("campaignDetails", [])
        print("Campaigns:", len(campaigns))
        for entry in campaigns:
            campaign = entry.get("campaign", {})
            campaign_name = campaign.get("name", "Unknown")
            campaign_id = campaign.get("id", "")

            ad_groups = campaign.get("adGroups", [])  # ✅ correct location
            print(f"\nCampaign: {campaign_name}")
            print("AdGroups found:", len(ad_groups))

            for ag in ad_groups:
                for criteria in ag.get("adGroupCriteria", []):
                    kw_criterion = criteria.get("keywordCriterion")
                    if not kw_criterion:
                        continue
                    keyword_infos = kw_criterion.get("keywordInfos", [])
                    for kw in keyword_infos:
                        bid = kw.get("bid") or {}
                        units = int(bid.get("units", 0))
                        nanos = bid.get("nanos", 0)
                        bid_value = units + nanos / 1_000_000
                        results.append({
                            "campaign_id": campaign_id,
                            "campaign_name": campaign_name,
                            "keyword": kw.get("text"),
                            "match_type": kw.get("keywordMatchType"),
                            "bid_inr": round(bid_value, 4),
                        })
        return results

    keyword_bids = extract_keyword_bids(data)

    for kb in keyword_bids:
        print(f"[{kb['campaign_name']}]  {kb['keyword']:<30}  {kb['match_type']:<30}  ₹{kb['bid_inr']}")

    try:
        print("Connecting with Voylla DB at:",
              datetime.datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S'))

        conn = psycopg2.connect(**Voylla_config)
        cursor = conn.cursor()

        print("Voylla DB Connected at:",
              datetime.datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S'))

    except psycopg2.DatabaseError as error:
        body = f'Error connecting to db: {str(error)}'
        print(body)


    for kb in keyword_bids:
        cursor.execute("""
            INSERT INTO voylla.instamart_bid 
            (campaign_id, campaign_name, targeting, match_type, current_bid,"Brand")
            VALUES (%s, %s, %s, %s, %s,%s)
            ON CONFLICT (campaign_id, targeting, match_type,"Brand")
            DO UPDATE SET 
                current_bid = EXCLUDED.current_bid;
        """, (
            kb["campaign_id"],
            kb["campaign_name"],
            kb["keyword"],
            kb["match_type"],
            kb["bid_inr"],
            brand
        ))

    conn.commit()
    print("✅ Keyword bids inserted/updated successfully")
    
    
    def send_alert_email(subject_suffix, body_message, brand):
        try:
            # 🔁 Brand-wise recipients
            brand_config = {
                "Voylla": {
                    "to": ["amit.singh@voylla.com"],
                    "cc": ["amit.singh@voylla.com"]
                }
            }

            # Default fallback
            config = brand_config.get(brand, {
                "to": receiver_email if isinstance(receiver_email, list) else [receiver_email]
            })

            with yagmail.SMTP(Alert_email, Alert_password) as yag:
                subject = f"{SD['Table Name']} - {subject_suffix}"

                yag.send(
                    to=config["to"],
                    subject=subject,
                    contents=body_message,
                    headers={"From": f"{Alert_sender_name} <{Alert_email}>"}
                )

            print(f"Alert email sent for {brand}")

        except Exception as e:
            print(f"Failed to send email alert: {e}")


    def format_changes(df_filtered):
        # ✅ Deduplicate keywords first
        df_unique = (
            df_filtered
            .groupby("targeting", as_index=False)
            .agg({
                "current_cpm": "mean",     # or 'first'
                "Updated_CPM": "max"       # same logic as payload
            })
        )

        # Optional: sort by biggest change
        df_unique["diff"] = df_unique["Updated_CPM"] - df_unique["current_cpm"]
        df_unique = df_unique.sort_values(by="diff", key=abs, ascending=False)

        lines = []

        for _, row in df_unique.iterrows():
            before = int(row["current_cpm"])
            after = int(row["Updated_CPM"])
            keyword = row["targeting"]

            change = after - before
            direction = "⬆️" if change > 0 else "⬇️" if change < 0 else "➡️"

            lines.append(f"{keyword}\n   {before} → {after} ({direction} {change})")

        return "\n\n".join(lines)

    # HOW CPM → nanos works:
    #   Instamart stores bids as  { units: int, nanos: int }
    #   e.g.  CPM = 139  →  units = 139, nanos = 0
    #         CPM = 20.5 →  units = 20,  nanos = 500_000_000
    #   Formula used here (integer CPMs only, matching your DB):
    #       units = int(cpm)
    #       nanos = 0
    
    import uuid
    from datetime import datetime, timezone

    def cpm_to_bid(cpm: float) -> dict:
        """Convert a CPM float to Instamart's { currency_code, units, nanos } bid format."""
        value = cpm / 1000  # convert CPM → per impression

        units = int(value)
        nanos = round((value - units) * 1_000_000_000)

        return {
            "currency_code": "INR",
            "units": units,
            "nanos": nanos
        }


    def update_instamart_status(campaign_id, brand, config):
        """Mark all rows for this campaign on the latest action_date as done."""
        try:
            conn_upd = psycopg2.connect(**config)
            cur_upd = conn_upd.cursor()
            cur_upd.execute(
                """
                UPDATE voylla."Instamart_actions_llm"
                SET update_status = 'done'
                WHERE campaign_id = %s
                  AND "brand" = %s
                  AND action_date = (
                      SELECT MAX(action_date)
                      FROM voylla."Instamart_actions_llm"
                      WHERE campaign_id = %s AND "brand" = %s
                  )
                """,
                (str(campaign_id), brand, str(campaign_id), brand)
            )
            conn_upd.commit()
            cur_upd.close()
            conn_upd.close()
            print(f"  update_status = 'done' saved for campaign {campaign_id}")
        except Exception as e:
            print(f"  Failed to update update_status for {campaign_id}: {e}")

    query_instamart = f"""
WITH touched_campaigns AS (
    SELECT DISTINCT campaign_id
    FROM voylla."Instamart_actions_llm"
    WHERE "brand" = '{brand}'
      AND user_implemented =  TRUE
      AND action_date = (
          SELECT MAX(action_date)
          FROM voylla."Instamart_actions_llm"
          WHERE "brand" = '{brand}'
      )
),

latest_actions AS (
    SELECT ins.action_date, ib.campaign_id, ib.campaign_name, ib.current_bid, ib.min_bid, ins.targeting, ins."action",
    ins.user_implemented, ins.cpm_change, ins.override_note, ins.brand, ins.update_status
    FROM voylla."Instamart_actions_llm" ins
    LEFT JOIN {brand}.instamart_bid ib 
    ON ins.campaign_id::TEXT = ib.campaign_id::TEXT AND REPLACE(ins.targeting,'_',' ')  = ib.targeting
    AND ins.match_type = ib.match_type
    WHERE "brand" = '{brand}'
      AND action_date = (
          SELECT MAX(action_date)
          FROM voylla."Instamart_actions_llm"
          WHERE "brand" = '{brand}'
      )
      AND ib.campaign_id IN (SELECT campaign_id FROM touched_campaigns)
),

updated_keywords AS (
    SELECT
        campaign_id,
        REPLACE(targeting, '_', ' ') AS targeting,
        CASE
            WHEN "action" LIKE 'INCREASE%'
                THEN (current_bid::FLOAT * (1 + 10::FLOAT / 100))::INT
            WHEN "action" LIKE 'DECREASE%'
                THEN (current_bid::FLOAT * (1 - 10::FLOAT / 100))::INT
            ELSE current_bid
        END AS updated_cpm
    FROM latest_actions
    WHERE user_implemented =  TRUE
)

SELECT
    la.action_date,
    la.campaign_id,
    la.campaign_name,
    REPLACE(la.targeting, '_', ' ')          AS targeting,
    la."action",
    la.cpm_change,
    la.user_implemented,
    la.override_note,
    la.current_bid as "current_cpm",
    COALESCE(uk.updated_cpm, la.current_bid) AS "Updated_CPM",
    la.brand,update_status
FROM latest_actions la
LEFT JOIN updated_keywords uk
    ON  la.campaign_id = uk.campaign_id
    AND REPLACE(la.targeting, '_', ' ') = uk.targeting
WHERE la.current_bid <> 0
AND la.action <> 'PAUSE'
AND (la.update_status IS NULL OR la.update_status != 'done')
ORDER BY la.action_date DESC;
    """

    try:
        print("\n" + "─" * 60)
        print(f"[Instamart] Query started at: "
              f"{datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S')}")

        conn   = psycopg2.connect(**Voylla_config)
        cursor = conn.cursor()
        cursor.execute(query_instamart)

        print(f"[Instamart] Query completed at: "
              f"{datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S')}")

        data   = cursor.fetchall()
        cols   = [col[0] for col in cursor.description]
        df     = pd.DataFrame(data, columns=cols)

        if df.empty:
            print(f"[Instamart] No actions found for {brand} — skipping Instamart.")
            with yagmail.SMTP(Alert_email, Alert_password) as yag:
                yag.send(
                    to=receiver_email,
                    subject=f"{SD['Script_Name']} - [Instamart] No data for {brand}",
                    contents=f"Instamart query returned 0 rows for brand: {brand}",
                    headers={"From": f"{Alert_sender_name} <{Alert_email}>"}
                )
        else:
            print(f"[Instamart] Rows fetched: {len(df)}")

    except psycopg2.Error as error:
        print(f"[Instamart] DB error: {error}")
        with yagmail.SMTP(Alert_email, Alert_password) as yag:
            yag.send(
                to=receiver_email,
                subject=f"{SD['Script_Name']} - [Instamart] Query Error",
                contents=f"Error executing Instamart SQL query:\n\n{error}",
                headers={"From": f"{Alert_sender_name} <{Alert_email}>"}
            )
        df = pd.DataFrame()   # ensure loop below is skipped safely

    finally:
        cursor.close()
        conn.close()
        
    print("Updating CPM .....")
    # ── 2. Loop over each campaign and fire PUT ────────────────────────────────
    for campaign_id in df['campaign_id'].unique():
        time.sleep(10)

        df_filtered = df[df['campaign_id'] == campaign_id]
        campaign_name = df_filtered['campaign_name'].iloc[0]

        # ── Deduplicate keywords (take max Updated_CPM per keyword) ──────────
        df_keywords = (
            df_filtered
            .groupby('targeting', as_index=False)
            .agg({'Updated_CPM': 'max'})
        )

        # ── Build keyword_infos list ──────────────────────────────────────────
        keyword_infos = []
        for _, krow in df_keywords.iterrows():
            keyword_infos.append({
                "text": krow["targeting"],
                "keyword_match_type": "KEYWORD_MATCH_TYPE_BROAD",   # adjust if needed
                "bid": cpm_to_bid(krow["Updated_CPM"])
            })

        im_payload = {
            "account_id": ad_account,          
            "campaign_id": campaign_id,
            "operation": [
                {
                    "update_live_campaign_operation": {
                        "update_ad_group_criterion": {
                            "ad_group_id": f"{campaign_id}#ag1",
                            "ad_group_criterion": {
                                "keyword_criterion": {
                                    "keyword_infos": keyword_infos
                                }
                            }
                        }
                    }
                }
            ],
            "request_context": {
                "request_id": str(uuid.uuid4()),
                "request_time": datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ"),
                "client_id": "IM_ADS_EXTERNAL_DASHBOARD"
            },
            "updated_by": {
                "email_id": "nitish.sharma@voylla.com"        
            }
        }

        im_url = "https://brand-portal-service-http.swiggy.com/api/v1/campaign"

        im_headers = {
            "accept": "application/json, text/plain, */*",
            "content-type": "application/json",
            "origin": "https://partner.instamart.in",
            "referer": f"https://partner.instamart.in/instamart/campaign/{campaign_id}/edit-1cc",
            "user-agent": (
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36"
            ),
            "Authorization": authorization,   
        }

        try:
            im_response = requests.put(im_url, json=im_payload, headers=im_headers)

            print(f"\n[Instamart] Campaign: {campaign_id}")
            print(f"[Instamart] Status  : {im_response.status_code}")
            print(f"[Instamart] Response: {im_response.text[:300]}")   # trim long responses

            # ── SUCCESS ──────────────────────────────────────────────────────
            if im_response.status_code in [200, 201]:

                changes_text = format_changes(df_filtered)

                body = f"""
                Hi Team,

                ✅ [Instamart] Campaign Update Successful

                Campaign ID   : {campaign_id}
                Campaign Name : {campaign_name}

                📊 Bid Changes:

                {changes_text}
                """
                send_alert_email("IM - Campaign Success", body, brand)
                update_instamart_status(campaign_id, brand, Voylla_config)

            # ── FAILURE ───────────────────────────────────────────────────────
            else:
                body = f"""
                Hi Team,

                ❌ [Instamart] Campaign Update Failed

                Campaign ID   : {campaign_id}
                Campaign Name : {campaign_name}

                Status Code: {im_response.status_code}

                Response:
                {im_response.text}
                """
                send_alert_email("IM - Campaign Failed", body, brand)

        except Exception as e:
            print(f"[Instamart] ❌ Exception for campaign {campaign_id}: {e}")

            body = f"""
            Hi Team,

            🚨 [Instamart] Exception while updating campaign

            Campaign ID   : {campaign_id}
            Campaign Name : {campaign_name}

            Error:
            {str(e)}
            """
            send_alert_email("IM - Campaign Error", body, brand)

        print("[Instamart] ------")

```
