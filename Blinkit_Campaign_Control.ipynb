```python
import requests
import pandas as pd
import time, random, json, traceback
from sqlalchemy import create_engine

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import imaplib
import email
from bs4 import BeautifulSoup
from datetime import datetime, timedelta
import datetime as dt_module


# =========================
# HELPERS
# =========================

def retry_call(func, retries=3, delay=5, label=""):
    last_exc = None
    for attempt in range(1, retries + 1):
        try:
            return func()
        except Exception as exc:
            last_exc = exc
            print(f"  [Retry {attempt}/{retries}] {label}: {exc}")
            if attempt < retries:
                time.sleep(delay)
    raise Exception(f"All {retries} retries failed — {label}: {last_exc}")


def human_typing_advanced(element, text):
    element.clear()
    for i, char in enumerate(text):
        element.send_keys(char)
        if i % random.randint(4, 7) == 0:
            time.sleep(random.uniform(0.2, 0.4))
        else:
            time.sleep(random.uniform(0.06, 0.15))


def extract_firebase_token(driver):
    logs = driver.get_log("performance")
    for entry in logs:
        message = json.loads(entry["message"])["message"]
        if message["method"] == "Network.requestWillBeSent":
            headers = message.get("params", {}).get("request", {}).get("headers", {})
            if "firebase_user_token" in headers:
                return headers["firebase_user_token"]
    return None


def extract_sign_in_link(email_id, password, sender, to_email):
    IMAP_SERVER = "imap.gmail.com"
    mail = imaplib.IMAP4_SSL(IMAP_SERVER)
    mail.login(email_id, password)
    mail.select("INBOX")

    since_date = (datetime.now() - timedelta(days=1)).strftime("%d-%b-%Y")
    search_criteria = (
        f'(FROM "{sender}" TO "{to_email}" '
        f'SUBJECT "Blinkit Brand Central" SINCE "{since_date}")'
    )

    status, messages = mail.search(None, search_criteria)
    if status != "OK":
        mail.logout()
        raise Exception("Email search failed")

    email_ids = messages[0].split()
    if not email_ids:
        mail.logout()
        raise Exception("No email found")

    status, msg_data = mail.fetch(email_ids[-1], "(RFC822)")
    msg = email.message_from_bytes(msg_data[0][1])

    for part in msg.walk():
        if part.get_content_type() == "text/html":
            html = part.get_payload(decode=True)
            soup = BeautifulSoup(html.decode(errors="ignore"), "html.parser")
            for a in soup.find_all("a", href=True):
                if "sign in to your account" in a.get_text(strip=True).lower():
                    mail.logout()
                    return a["href"]

    mail.logout()
    raise Exception("Sign-in link not found")


# =========================
# LOGIN FUNCTION (RETRY TARGET)
# =========================

def login_and_get_token(email_id, password, sender, ad_account):

    options = Options()
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.set_capability("goog:loggingPrefs", {"performance": "ALL"})

    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Network.enable", {})

    try:
        driver.get("https://brands.blinkit.com/")
        time.sleep(3)

        wait = WebDriverWait(driver, 20)

        wait.until(
            EC.element_to_be_clickable((By.XPATH, "//button[.//div[text()='Login']]"))
        ).click()

        username = wait.until(
            EC.visibility_of_element_located((By.CSS_SELECTOR, "input[placeholder='Email']"))
        )
        human_typing_advanced(username, ad_account)

        wait.until(
            EC.element_to_be_clickable((By.XPATH, "//span[text()='Request Sign in Link']"))
        ).click()

        print("Waiting for email...")
        time.sleep(15)

        # Email retry
        def get_link():
            return extract_sign_in_link(email_id, password, sender, ad_account)

        link = retry_call(get_link, retries=3, delay=10, label="Email fetch")
        print("Login link fetched")

        driver.get(link)
        time.sleep(3)

        driver.get("https://brands.blinkit.com/diy/list")
        time.sleep(3)

        token = extract_firebase_token(driver)

        if not token:
            raise Exception("Token not found")

        print("Token captured ✅")
        return driver, token

    except Exception as e:
        driver.quit()
        raise e


# =========================
# DB CONNECTION
# =========================

db_connection_str = (
    f"postgresql+psycopg2://{Voylla_config['user']}:{Voylla_config['password']}"
    f"@{Voylla_config['host']}:{Voylla_config['port']}/{Voylla_config['database']}"
)
engine = create_engine(db_connection_str)

df_brands = retry_call(lambda: pd.read_sql(query, engine), label="fetch df_brands")


# =========================
# MAIN EXECUTION
# =========================

all_rows = []

for _, row in df_brands.iterrows():

    brand      = row['brand']
    email_id   = row['email']
    password   = row['password']
    ad_account = row['ad_account_id']
    sender     = row['Sender']

    print(f"\n{'='*60}")
    print(f"Running for Brand: {brand}")
    print(f"{'='*60}")

    # -------------------------
    # CAMPAIGNS
    # -------------------------
    campaign_query = f"""
        SELECT DISTINCT "Campaign ID"
        FROM voylla."Blinkit_Ads_Report"
        WHERE TO_TIMESTAMP("Date",'YYYY-MM-DD HH24:MI:SS')
        >= CURRENT_DATE - INTERVAL '7 days'
        AND "Brand" = '{brand}'
    """

    try:
        campaign_df = retry_call(lambda: pd.read_sql(campaign_query, engine),
                                label=f"campaign fetch {brand}")
        campaign_ids = campaign_df["Campaign ID"].tolist()
    except Exception as e:
        print(f"Campaign fetch failed: {e}")
        continue

    if not campaign_ids:
        print("No campaigns found")
        continue

    print(f"Campaigns: {len(campaign_ids)}")

    # -------------------------
    # LOGIN WITH RETRY
    # -------------------------
    try:
        driver, token = retry_call(
            lambda: login_and_get_token(email_id, password, sender, ad_account),
            retries=3,
            delay=10,
            label=f"Login {brand}"
        )
    except Exception as e:
        print(f"Login failed after retries: {e}")
        continue

    # -------------------------
    # API FETCH
    # -------------------------
    for campaign_id in campaign_ids:

        def fetch_campaign():
            url = f"https://brands.blinkit.com/adservice/v1/campaigns/{campaign_id}"
            headers = {
                "accept": "application/json",
                "user-agent": "Mozilla/5.0",
                "referer": f"https://brands.blinkit.com/diy/campaign/{campaign_id}",
                "firebase_user_token": token
            }
            r = requests.get(url, headers=headers, timeout=30)
            if r.status_code != 200:
                raise Exception(f"{r.status_code} {r.text[:200]}")
            return r.json()

        try:
            data = retry_call(fetch_campaign, label=f"Campaign {campaign_id}")
        except Exception as e:
            print(f"Skipping {campaign_id}: {e}")
            continue

        campaign_budget = data.get("data", {}).get("campaign", {}).get("campaign_budget")

        for item in data.get("data", {}).get("campaign", {}).get("keywords", []):
            keyword = item.get("keyword")
            for bid in item.get("bids", []):
                all_rows.append({
                    "Brand": brand,
                    "campaign_id": campaign_id,
                    "keyword": keyword,
                    "match_type": bid.get("match_type"),
                    "cpm": bid.get("cpm"),
                    "max_boost": bid.get("max_boost"),
                    "budget": campaign_budget
                })

    driver.quit()


# =========================
# FINAL OUTPUT
# =========================

final_df = pd.DataFrame(all_rows)

current_hour = dt_module.datetime.now().hour

if current_hour < 12:
    try:
        existing_budget_df = retry_call(
            lambda: pd.read_sql(
                'SELECT campaign_id, MAX(budget) AS budget FROM voylla."Blinkit_CPM" GROUP BY campaign_id',
                engine
            )
        )

        budget_map = dict(zip(
            existing_budget_df['campaign_id'].astype(str),
            existing_budget_df['budget']
        ))

        final_df['budget'] = (
            final_df['campaign_id'].astype(str)
            .map(budget_map)
            .combine_first(final_df['budget'])
        )

        print(f"[{current_hour}:xx] Budget preserved from DB")

    except Exception as e:
        print(f"Budget protection failed: {e}")

else:
    print(f"[{current_hour}:xx] Fresh budget saved")

print("\nFINAL DATA")
print(final_df.head())
```

```python

# ── Helper functions ──────────────────────────────────────────────────────────

def format_changes(df_filtered):
    df_unique = (
        df_filtered
        .groupby("targeting", as_index=False)
        .agg({"current_cpm": "mean", "Updated_CPM": "max"})
    )
    df_unique["current_cpm"] = pd.to_numeric(df_unique["current_cpm"], errors="coerce").fillna(0)
    df_unique["Updated_CPM"] = pd.to_numeric(df_unique["Updated_CPM"], errors="coerce").fillna(0)
    df_unique["diff"] = df_unique["Updated_CPM"] - df_unique["current_cpm"]
    df_unique = df_unique.sort_values(by="diff", key=abs, ascending=False)
    lines = []
    for _, row in df_unique.iterrows():
        before    = int(row["current_cpm"])
        after     = int(row["Updated_CPM"])
        keyword   = row["targeting"]
        change    = after - before
        direction = "⬆️" if change > 0 else "⬇️" if change < 0 else "➡️"
        lines.append(f"{keyword}\n   {before} → {after} ({direction} {change})")
    return "\n\n".join(lines)


def human_typing_advanced(element, text):
    element.clear()
    for i, char in enumerate(text):
        element.send_keys(char)
        if i % random.randint(4, 7) == 0:
            time.sleep(random.uniform(0.2, 0.4))
        else:
            time.sleep(random.uniform(0.06, 0.15))


def extract_firebase_token(driver):
    logs = driver.get_log("performance")
    for entry in logs:
        message = json.loads(entry["message"])["message"]
        if message["method"] == "Network.requestWillBeSent":
            hdrs = message.get("params", {}).get("request", {}).get("headers", {})
            if "firebase_user_token" in hdrs:
                return hdrs["firebase_user_token"]
    return None

# ,"gaurav.ajmera@voylla.com"
def send_alert_email(subject_suffix, body_message, brand):
    try:
        brand_config = {
            "Voylla":  {"to": ["amit.singh@voylla.com","gaurav.ajmera@voylla.com"],             "cc": ["amit.singh@voylla.com"]},
            "Petcrux": {"to": ["adyasha.behera@goatbrandlabs.com"],  "cc": ["amit.singh@voylla.com"]}
        }
        config = brand_config.get(brand, {
            "to": receiver_email if isinstance(receiver_email, list) else [receiver_email],
            "cc": []
        })
        table_name = SD['Table Name'].iloc[0] if isinstance(SD, pd.DataFrame) else SD['Table Name']
        with yagmail.SMTP(Alert_email, Alert_password) as yag:
            subject = f"{table_name} - {subject_suffix}"
            yag.send(to=config["to"], cc=config["cc"], subject=subject,
                     contents=body_message,
                     headers={"From": f"{Alert_sender_name} <{Alert_email}>"})
        print(f"Alert email sent for {brand}")
    except Exception as e:
        print(f"Failed to send email alert: {e}")


# ── DB helpers for slot status ─────────────────────────────────────────────────

def update_slot_status(campaign_id, slot, status, conn):
    """Write DONE / FAILED / SKIPPED + timestamp into schedule table."""
    col_status   = f"{slot}_status"
    col_executed = f"{slot}_executed_at"
    sql = f"""
        UPDATE voylla."Blinkit_Campaign_Schedule"
        SET   "{col_status}"   = %s,
              "{col_executed}" = NOW(),
              updated_at       = NOW()
        WHERE campaign_id = %s
    """
    try:
        with conn.cursor() as cur:
            cur.execute(sql, (status, str(campaign_id)))
        conn.commit()
        print(f"  [{slot}] status → {status} for campaign {campaign_id}")
    except Exception as e:
        print(f"  Failed to update slot status: {e}")
        conn.rollback()


def reset_slot_if_new_day(campaign_id, slot, executed_at, conn):
    """Reset slot to PENDING if last execution was on a previous day."""
    import datetime as _dt
    today = _dt.date.today()
    if executed_at is None or pd.isna(executed_at):
        return   # never ran — already PENDING
    try:
        exec_date = pd.Timestamp(executed_at).date()
        if exec_date < today:
            col = f"{slot}_status"
            sql = f"""
                UPDATE voylla."Blinkit_Campaign_Schedule"
                SET "{col}" = 'PENDING', updated_at = NOW()
                WHERE campaign_id = %s
            """
            with conn.cursor() as cur:
                cur.execute(sql, (str(campaign_id),))
            conn.commit()
            print(f"  [{slot}] auto-reset to PENDING (last run was {exec_date})")
    except Exception as e:
        print(f"  Failed daily reset: {e}")
        conn.rollback()


def in_time_window(start_t, end_t, now_t):
    """True if now_t is within [start_t, end_t).
    Handles None, timedelta (psycopg2), and midnight-crossing slots (e.g. 22:00-00:00).
    """
    if start_t is None or end_t is None:
        return False
    try:
        if pd.isna(start_t) or pd.isna(end_t):
            return False
    except Exception:
        pass
    import datetime as _dt
    # Convert timedelta → time (psycopg2 sometimes returns timedelta for TIME columns)
    if isinstance(start_t, _dt.timedelta):
        start_t = (_dt.datetime.min + start_t).time()
    if isinstance(end_t, _dt.timedelta):
        end_t = (_dt.datetime.min + end_t).time()
    # Midnight-crossing slot: end_t is 00:00 or end_t < start_t (e.g. 22:00 → 00:00 or 23:00 → 05:00)
    midnight = _dt.time(0, 0, 0)
    # start == end means slot not configured (e.g. 00:00-00:00 or 02:00-02:00)
    if start_t == end_t:
        return False
    if end_t == midnight or end_t < start_t:
        # Slot wraps past midnight: active if now >= start OR now < end
        # (end==00:00 means "until end of day", so any time >= start qualifies)
        return now_t >= start_t or (end_t != midnight and now_t < end_t)
    return start_t <= now_t < end_t


# ── DB + brand loop ───────────────────────────────────────────────────────────

db_connection_str = (
    f"postgresql+psycopg2://{Voylla_config['user']}:{Voylla_config['password']}"
    f"@{Voylla_config['host']}:{Voylla_config['port']}/{Voylla_config['database']}"
)
db_connection = create_engine(db_connection_str)

df_brands = retry_call(lambda: pd.read_sql(query, db_connection),
                       label="fetch df_brands for restart loop")

for index, row in df_brands.iterrows():
    brand      = row['brand']
    email_1    = row['email']
    password_1 = row['password']
    ad_account = row['ad_account_id']
    sender     = row['Sender']

    print(f"\n{'='*70}")
    print(f"Processing brand: {brand}")
    print(f"{'='*70}\n")

    # ── Read schedule table for this brand ────────────────────────────────────
    schedule_query = f"""
        SELECT *
        FROM voylla."Blinkit_Campaign_Schedule"
        WHERE brand = '{brand}'
    """
    try:
        df_schedule = retry_call(lambda: pd.read_sql(schedule_query, db_connection),
                                 label=f"fetch schedule for {brand}")
    except Exception as e:
        print(f"Could not read schedule for {brand}: {e} — skipping brand")
        continue

    if df_schedule.empty:
        print(f"No schedule configured for {brand} — skipping")
        continue

    scheduled_ids   = df_schedule['campaign_id'].astype(str).tolist()
    ids_in_clause   = ", ".join([f"'{c}'" for c in scheduled_ids])
    print(f"Scheduled campaigns: {scheduled_ids}")

    # ── Selenium login ─────────────────────────────────────────────────────────
    options = Options()
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    options.add_argument(
        "user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36"
    )
    options.set_capability("goog:loggingPrefs", {"performance": "ALL"})

    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Network.enable", {})

    token = _cfuvid = __cfruid = __cf_bm = None
    SUBJECT_KEYWORD = "Blinkit Brand Central"
    TO_EMAIL = ad_account

    LOGIN_ATTEMPTS = 3
    for login_attempt in range(1, LOGIN_ATTEMPTS + 1):
        try:
            driver.get("https://brands.blinkit.com/")
            time.sleep(2)
            driver.get("https://brands.blinkit.com/")
            time.sleep(5)

            wait = WebDriverWait(driver, 20)
            login_btn = WebDriverWait(driver, 10).until(
                EC.element_to_be_clickable((By.XPATH, "//button[.//div[text()='Login']]"))
            )
            login_btn.click()
            time.sleep(2)

            username_field = wait.until(
                EC.visibility_of_element_located((By.CSS_SELECTOR, "input[placeholder='Email']"))
            )
            human_typing_advanced(username_field, ad_account)

            request_link = WebDriverWait(driver, 10).until(
                EC.element_to_be_clickable((By.XPATH, "//span[text()='Request Sign in Link']"))
            )
            request_link.click()
            time.sleep(15)

            import imaplib, email as _email_lib
            from bs4 import BeautifulSoup
            from datetime import datetime as _dt_cls, timedelta

            IMAP_SERVER = "imap.gmail.com"

            def extract_sign_in_link():
                mail = imaplib.IMAP4_SSL(IMAP_SERVER)
                mail.login(email_1, password_1)
                mail.select("INBOX")
                since_date = (_dt_cls.now() - timedelta(days=1)).strftime("%d-%b-%Y")
                criteria   = (f'(FROM "{sender}" TO "{TO_EMAIL}" '
                              f'SUBJECT "{SUBJECT_KEYWORD}" SINCE "{since_date}")')
                status, messages = mail.search(None, criteria)
                if status != "OK":
                    mail.logout(); raise Exception("Email search failed")
                ids = messages[0].split()
                if not ids:
                    mail.logout(); raise Exception("No matching email found")
                status, msg_data = mail.fetch(ids[-1], "(RFC822)")
                if status != "OK":
                    mail.logout(); raise Exception("Failed to fetch email")
                msg  = _email_lib.message_from_bytes(msg_data[0][1])
                link = None
                for part in msg.walk():
                    if part.get_content_type() == "text/html":
                        html = part.get_payload(decode=True)
                        if not html: continue
                        soup = BeautifulSoup(html.decode(errors="ignore"), "html.parser")
                        for a in soup.find_all("a", href=True):
                            if "sign in to your account" in a.get_text(strip=True).lower():
                                link = a["href"]; break
                    if link: break
                mail.logout()
                if not link: raise Exception("Sign-in link not found")
                return link

            sign_in_url = retry_call(extract_sign_in_link, retries=3, delay=10,
                                     label="extract sign-in link")
            print("Sign-in link extracted")

            driver.get(sign_in_url)
            time.sleep(2)
            driver.get("https://brands.blinkit.com/diy/list")
            time.sleep(2)

            cookies_dict = {c['name']: c['value'] for c in driver.get_cookies()}
            _cfuvid  = cookies_dict.get("_cfuvid")
            __cfruid = cookies_dict.get("__cfruid")
            __cf_bm  = cookies_dict.get("__cf_bm")

            token = extract_firebase_token(driver)
            if token:
                print("Firebase token captured")
                break
            else:
                print(f"Token not found (attempt {login_attempt}/{LOGIN_ATTEMPTS})")
                time.sleep(50)

        except Exception as e:
            print(f"Login attempt {login_attempt} failed: {e}")
            traceback.print_exc()

    if not token:
        print(f"Could not obtain token for {brand} — skipping")
        driver.quit()
        continue

    # ── Fetch CPM / keyword data for all scheduled campaigns ──────────────────
    query3 = f"""

            SELECT
            la.campaign_id,
            la.campaign_name,
            TO_CHAR(la.start_ts + INTERVAL '1 day', 'MM/DD/YYYY') AS start_ts,
            TO_CHAR(la.end_ts, 'MM/DD/YYYY')                      AS end_ts,
            la.campaign_type,
            la.objective_type,
            la.product_id,
            REPLACE(bc.keyword,'_',' ') AS targeting,
            bc.cpm AS current_cpm,
            bc.cpm AS "Updated_CPM",
            region_type,
				CASE 
				    WHEN TRANSLATE(region_ids::TEXT, '{{}}', '') = '' 
				    THEN '-1' 
				    ELSE TRANSLATE(region_ids::TEXT, '{{}}', '')
				END AS region_ids,
            bc."Brand"
        FROM Voylla."Blinkit_CampaignWise_ProductID" la
        LEFT JOIN Voylla."Blinkit_CPM" bc
            ON  la.campaign_id::TEXT = bc.campaign_id::TEXT
        WHERE (bc.cpm <> 0 OR bc.cpm IS NOT NULL OR bc.cpm::TEXT != '' )
          AND la.campaign_id IN ({ids_in_clause})



        




    
    """

    import datetime as _datetime
    conn_status = None
    try:
        print("Query started at: " + _datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S'))

        df_campaign_details = None
        for attempt in range(1, MAX_RETRIES + 1):
            try:
                conn_misauto = psycopg2.connect(**Voylla_config)
                cursor       = conn_misauto.cursor()
                cursor.execute(query3)
                break
            except psycopg2.Error as db_err:
                print(f"DB query attempt {attempt}/{MAX_RETRIES} failed: {db_err}")
                if attempt < MAX_RETRIES: time.sleep(RETRY_DELAY)
                else: raise

        data   = cursor.fetchall()
        column = [col[0] for col in cursor.description]
        df_campaign_details = pd.DataFrame(data, columns=column)
        print(f"Query done — {len(df_campaign_details)} rows at: "
              + _datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S'))

        if df_campaign_details.empty:
            script_name = SD['Script_Name'].iloc[0] if isinstance(SD, pd.DataFrame) else SD['Script_Name']
            with yagmail.SMTP(Alert_email, Alert_password) as yag:
                subject = f"{script_name} - No CPM data for {brand}"
                body    = f"Hi Team,\n\nNo CPM data returned for {brand}.\n\nQuery:\n{query3}"
                yag.send(to=receiver_email, subject=subject, contents=body,
                         headers={"From": f"{Alert_sender_name} <{Alert_email}>"})
            driver.quit()
            continue

    except psycopg2.Error as error:
        script_name = SD['Script_Name'].iloc[0] if isinstance(SD, pd.DataFrame) else SD['Script_Name']
        body = f'Error executing SQL query: {str(error)}'
        print(body)
        with yagmail.SMTP(Alert_email, Alert_password) as yag:
            subject = script_name + " - Query Error"
            yag.send(to=receiver_email, cc=None, bcc=None, subject=subject, contents=body,
                     headers={"From": f"{Alert_sender_name} <{Alert_email}>"})
        driver.quit()
        continue

    finally:
        try:
            cursor.close()
        except Exception:
            pass

    # Open a persistent connection for writing slot statuses
    try:
        conn_status = psycopg2.connect(**Voylla_config)
    except Exception as e:
        print(f"Could not open status-update connection: {e}")
        conn_status = None

    # ── Time snapshot (once per brand run) ────────────────────────────────────
    now_dt       = _datetime.datetime.now()
    today_date   = now_dt.date()
    current_time = now_dt.time()

    # ── Process each scheduled campaign ───────────────────────────────────────
    for _, sched in df_schedule.iterrows():

        campaign_id   = str(sched['campaign_id'])
        campaign_id_i = sched['campaign_id']   # original type for payload

        print(f"\n--- Campaign {campaign_id} ---")

        # ── Daily auto-reset ─────────────────────────────────────────────────
        if conn_status:
            reset_slot_if_new_day(campaign_id, 'slot1', sched['slot1_executed_at'], conn_status)
            reset_slot_if_new_day(campaign_id, 'slot2', sched['slot2_executed_at'], conn_status)
            # Re-read fresh status after potential reset
            try:
                fresh = pd.read_sql(
                    f"""SELECT slot1_status, slot2_status
                          FROM voylla."Blinkit_Campaign_Schedule"
                          WHERE campaign_id = '{campaign_id}'""",
                    db_connection
                )
                if not fresh.empty:
                    sched = sched.copy()
                    sched['slot1_status'] = fresh['slot1_status'].iloc[0]
                    sched['slot2_status'] = fresh['slot2_status'].iloc[0]
            except Exception:
                pass

        # ── Determine current slot ────────────────────────────────────────────
        # Check slot1 first. Only check slot2 if slot1 does NOT match.
        # This avoids ambiguity when slot windows overlap.
        in_slot1 = in_time_window(sched['slot1_start'], sched['slot1_end'], current_time)
        in_slot2 = (not in_slot1) and in_time_window(sched['slot2_start'], sched['slot2_end'], current_time)
        in_gap   = not in_slot1 and not in_slot2

        print(f"  Time: {current_time.strftime('%H:%M')} | "
              f"slot1={'✓' if in_slot1 else '✗'} | "
              f"slot2={'✓' if in_slot2 else '✗'} | "
              f"gap={'✓' if in_gap else '✗'}")

        # ── Gap: pause campaign ───────────────────────────────────────────────
        if in_gap:
            if sched.get('pause_on_gap', True):
                print(f"  [GAP] Pausing campaign {campaign_id}")
                try:
                    pause_url     = f"https://brands.blinkit.com/adservice/v1/campaigns/{campaign_id}"
                    pause_headers = {
                        "accept": "application/json, text/plain, */*",
                        "user-agent": "Mozilla/5.0",
                        "origin": "https://brands.blinkit.com",
                        "referer": f"https://brands.blinkit.com/diy/campaign/{campaign_id}",
                        "firebase_user_token": token
                    }
                    pause_cookies = {"__cf_bm": __cf_bm, "__cfruid": __cfruid, "_cfuvid": _cfuvid}

                    def do_pause_gap():
                        r = requests.delete(pause_url, headers=pause_headers,
                                            cookies=pause_cookies, timeout=30)
                        if r.status_code not in [200, 201, 204]:
                            raise Exception(f"HTTP {r.status_code}")
                        return r

                    retry_call(do_pause_gap, retries=3, delay=5,
                               label=f"pause (gap) campaign {campaign_id}")
                    send_alert_email(
                        "Campaign Paused (gap)",
                        f"Hi Team,\n\nCampaign {campaign_id} paused — outside all slot windows.",
                        brand
                    )
                except Exception as e:
                    print(f"  Pause failed: {e}")
            else:
                print(f"  [GAP] pause_on_gap=False — no action")
            continue

        # ── Determine active slot ─────────────────────────────────────────────
        if in_slot1:
            slot_label   = "slot1"
            slot_active  = bool(sched['slot1_active'])
            slot_status  = sched['slot1_status']
            total_budget = int(sched['slot1_budget']) if pd.notna(sched['slot1_budget']) and sched['slot1_budget'] else 500
        else:
            slot_label   = "slot2"
            slot_active  = bool(sched['slot2_active'])
            slot_status  = sched['slot2_status']
            total_budget = int(sched['slot2_budget']) if pd.notna(sched['slot2_budget']) and sched['slot2_budget'] else 500

        print(f"  Active slot: {slot_label} | accept={slot_active} | status={slot_status} | budget={total_budget}")

        # ── Slot disabled → pause ─────────────────────────────────────────────
        if not slot_active:
            print(f"  [{slot_label}] accept=False — pausing campaign")
            try:
                pause_url     = f"https://brands.blinkit.com/adservice/v1/campaigns/{campaign_id}"
                pause_headers = {
                    "accept": "application/json, text/plain, */*",
                    "user-agent": "Mozilla/5.0",
                    "origin": "https://brands.blinkit.com",
                    "referer": f"https://brands.blinkit.com/diy/campaign/{campaign_id}",
                    "firebase_user_token": token
                }
                pause_cookies = {"__cf_bm": __cf_bm, "__cfruid": __cfruid, "_cfuvid": _cfuvid}

                def do_pause_disabled():
                    r = requests.delete(pause_url, headers=pause_headers,
                                        cookies=pause_cookies, timeout=30)
                    if r.status_code not in [200, 201, 204]:
                        raise Exception(f"HTTP {r.status_code}")
                    return r

                retry_call(do_pause_disabled, retries=3, delay=5,
                           label=f"pause (disabled) campaign {campaign_id}")
                if conn_status:
                    update_slot_status(campaign_id, slot_label, 'SKIPPED', conn_status)
                send_alert_email(
                    f"Campaign Paused ({slot_label} disabled)",
                    f"Hi Team,\n\nCampaign {campaign_id} paused — {slot_label} accept=False.",
                    brand
                )
            except Exception as e:
                print(f"  Pause failed: {e}")
            continue

        # ── Already done today → skip ─────────────────────────────────────────
        if str(slot_status).upper() == 'DONE':
            print(f"  [{slot_label}] already DONE today — checking if campaign is active...")

            # Still verify campaign is live (belt-and-suspenders)
            try:
                def get_status():
                    r = requests.get(
                        f"https://brands.blinkit.com/adservice/v1/campaigns/{campaign_id}",
                        headers={"accept": "application/json", "user-agent": "Mozilla/5.0",
                                 "firebase_user_token": token,
                                 "referer": f"https://brands.blinkit.com/diy/campaign/{campaign_id}"},
                        cookies={"__cf_bm": __cf_bm, "__cfruid": __cfruid, "_cfuvid": _cfuvid},
                        timeout=30
                    )
                    if r.status_code != 200:
                        raise Exception(f"HTTP {r.status_code}")
                    return r.json()

                status_data    = retry_call(get_status, retries=2, delay=5,
                                            label=f"verify active {campaign_id}")
                camp_obj       = status_data.get("data", {}).get("campaign", {})
                current_status = (camp_obj.get("status") or camp_obj.get("campaign_status")
                                  or camp_obj.get("state") or "UNKNOWN")
                current_budget = camp_obj.get("campaign_budget")
                print(f"  API: status={current_status} | budget={current_budget}")

                if str(current_status).upper() == "ACTIVE" and int(current_budget or 0) == total_budget:
                    print(f"  Campaign confirmed ACTIVE with correct budget — no action needed")
                    continue
                else:
                    print(f"  Campaign not active or wrong budget — will re-restart")
                    # Fall through to restart below

            except Exception as e:
                print(f"  Could not verify status: {e} — skipping to be safe")
                continue

        # ── Get CPM data for this campaign ────────────────────────────────────
        df_filtered = df_campaign_details[
            df_campaign_details['campaign_id'].astype(str) == campaign_id
        ]

        if df_filtered.empty:
            print(f"  No CPM data found for campaign {campaign_id} — skipping")
            continue

        pid_list       = df_filtered['product_id'].astype(int).unique().tolist()
        asset_type     = df_filtered['campaign_type'].iloc[0]
        objective_type = df_filtered['objective_type'].iloc[0]
        start_ts       = df_filtered['start_ts'].iloc[0]
        end_ts         = df_filtered['end_ts'].iloc[0]
        campaign_name  = df_filtered['campaign_name'].iloc[0]
        region_ids     = df_filtered['region_ids'].iloc[0]

        # ── Build keyword payload ─────────────────────────────────────────────
        df_keywords = (
            df_filtered
            .groupby('targeting', as_index=False)
            .agg({'Updated_CPM': 'max'})
        )
        keywords_payload = [
            {
                "keyword": r["targeting"],
                "bids": [{"match_type": "EXACT",
                          "cpm": int(r["Updated_CPM"]),
                          "max_boost": None}]
            }
            for _, r in df_keywords.iterrows()
        ]

        req_headers = {
            "accept": "application/json, text/plain, */*",
            "content-type": "application/json;charset=UTF-8",
            "origin": "https://brands.blinkit.com",
            "referer": f"https://brands.blinkit.com/diy/edit/campaign/{campaign_id}",
            "user-agent": "Mozilla/5.0",
            "firebase_user_token": token,
        }
        req_cookies = {"__cf_bm": __cf_bm, "__cfruid": __cfruid, "_cfuvid": _cfuvid}

        payload = {
            "source_platform": "diy_dashboard_web",
            "requested_by": ad_account,
            "advertiser_id": 1073,
            "brand_name": "GBL Altair Private Limited",
            "campaign_id": campaign_id_i,
            "objective_type": objective_type,
            "asset_type": asset_type,
            "highlighted_pids": ",".join(map(str, pid_list)),
            "campaign_data": {
                "pids": ",".join(map(str, pid_list)),
                "products": [{"id": int(p), "is_star_product": False} for p in pid_list],
                "brand_ids": None,
                "category_ids": None
            },
            "campaign_start": start_ts,
            "campaign_end": end_ts,
            "infinite_campaign": True,
            "cpm": 0,
            "name": campaign_name,
            "bidding_strategy": {"total_budget": total_budget, "pacing_type": "DAILY"},
            "campaign_targeting": {
                "city_ids": str(region_ids),
                "is_extendable": False,
                "negative_keywords": [],
                "keyword_targeting": {"keywords": keywords_payload}
            },
            "campaign_request_type": "RESTART"
        }

        # ── Restart campaign ──────────────────────────────────────────────────
        def do_restart():
            r = requests.put("https://brands.blinkit.com/adservice/v3/campaigns",
                             json=payload, headers=req_headers,
                             cookies=req_cookies, timeout=30)
            print(f"  Restart status: {r.status_code}")
            if r.status_code not in [200, 201]:
                raise Exception(f"HTTP {r.status_code}: {r.text[:300]}")
            return r

        try:
            retry_call(do_restart, retries=3, delay=5,
                       label=f"restart campaign {campaign_id}")

            if conn_status:
                update_slot_status(campaign_id, slot_label, 'DONE', conn_status)

            changes_text = format_changes(df_filtered)
            body = f"""
Hi Team,

✅ Campaign Restarted Successfully

Campaign ID   : {campaign_id}
Campaign Name : {campaign_name}
Slot          : {slot_label} ({sched[f'{slot_label}_start']} – {sched[f'{slot_label}_end']})
Budget Applied: {total_budget}

📊 Bid Changes:

{changes_text}
"""
            send_alert_email(f"Campaign Success ({slot_label})", body, brand)

        except Exception as e:
            print(f"  Restart failed: {e}")
            if conn_status:
                update_slot_status(campaign_id, slot_label, 'FAILED', conn_status)
            body = f"""
Hi Team,

❌ Campaign Restart Failed (after {MAX_RETRIES} attempts)

Campaign ID   : {campaign_id}
Campaign Name : {campaign_name}
Slot          : {slot_label}
Budget        : {total_budget}

Error: {str(e)}
"""
            send_alert_email(f"Campaign Failed ({slot_label})", body, brand)

        print("------")

    # ── Cleanup ───────────────────────────────────────────────────────────────
    if conn_status:
        try:
            conn_status.close()
        except Exception:
            pass

    driver.quit()
```
