```python

# ── Helper functions (defined once, outside all loops) ────────────────────────

def format_changes(df_filtered):
    # ✅ Deduplicate keywords first
    df_unique = (
        df_filtered
        .groupby("targeting", as_index=False)
        .agg({
            "current_cpm": "mean",
            "Updated_CPM": "max"
        })
    )
    df_unique["diff"] = df_unique["Updated_CPM"] - df_unique["current_cpm"]
    df_unique = df_unique.sort_values(by="diff", key=abs, ascending=False)

    lines = []
    for _, row in df_unique.iterrows():
        before = int(row["current_cpm"])
        after = int(row["Updated_CPM"])
        keyword = row["targeting"]
        change = after - before
        direction = "⬆️" if change > 0 else "⬇️" if change < 0 else "➡️"
        lines.append(f"{keyword}\n   {before} → {after} ({direction} {change})")

    return "\n\n".join(lines)


def human_typing_advanced(element, text):
    element.clear()
    for i, char in enumerate(text):
        element.send_keys(char)
        if i % random.randint(4, 7) == 0:
            time.sleep(random.uniform(0.2, 0.4))
        else:
            time.sleep(random.uniform(0.06, 0.15))


def extract_firebase_token(driver):
    logs = driver.get_log("performance")
    for entry in logs:
        message = json.loads(entry["message"])["message"]
        if message["method"] == "Network.requestWillBeSent":
            hdrs = message.get("params", {}).get("request", {}).get("headers", {})
            if "firebase_user_token" in hdrs:
                return hdrs["firebase_user_token"]
    return None


def send_alert_email(subject_suffix, body_message, brand):
    try:
        brand_config = {
            "Voylla": {
                "to": ["amit.singh@voylla.com"],
                "cc": ["amit.singh@voylla.com"]
            },
            "Petcrux": {
                "to": ["adyasha.behera@goatbrandlabs.com"],
                "cc": ["amit.singh@voylla.com"]
            }
        }
        config = brand_config.get(brand, {
            "to": receiver_email if isinstance(receiver_email, list) else [receiver_email],
            "cc": []
        })
        table_name = SD['Table Name'].iloc[0] if isinstance(SD, pd.DataFrame) else SD['Table Name']
        with yagmail.SMTP(Alert_email, Alert_password) as yag:
            subject = f"{table_name} - {subject_suffix}"
            yag.send(
                to=config["to"],
                cc=config["cc"],
                subject=subject,
                contents=body_message,
                headers={"From": f"{Alert_sender_name} <{Alert_email}>"}
            )
        print(f"Alert email sent for {brand}")
    except Exception as e:
        print(f"Failed to send email alert: {e}")


def update_campaign_status(brand, campaign_id, status, voylla_config):
    """Update update_status in Blinkit_actions_llm for a campaign on its latest action_date."""
    try:
        conn = psycopg2.connect(**voylla_config)
        cur = conn.cursor()
        cur.execute(
            f"""
            UPDATE {brand}."Blinkit_actions_llm"
            SET update_status = %s
            WHERE "Brand" = %s
              AND campaign_id::TEXT = %s
              AND action_date::TEXT = (
                  SELECT to_char(date(MAX(action_date)) - INTERVAL '0' DAY, 'YYYY-MM-DD')
                  FROM {brand}."Blinkit_actions_llm"
                  WHERE "Brand" = %s
              )
            """,
            (status, brand, str(campaign_id), brand)
        )
        conn.commit()
        cur.close()
        conn.close()
        print(f"✅ update_status = '{status}' saved for campaign {campaign_id}")
    except Exception as e:
        print(f"❌ Failed to update update_status for campaign {campaign_id}: {e}")


db_connection_str = (
    f"postgresql+psycopg2://{Voylla_config['user']}:{Voylla_config['password']}"
    f"@{Voylla_config['host']}:{Voylla_config['port']}/{Voylla_config['database']}"
)
db_connection = create_engine(db_connection_str)
df_brands = pd.read_sql(query, db_connection)

for index, row in df_brands.iterrows():
    brand      = row['brand']
    email_1    = row['email']
    password_1 = row['password']
    ad_account = row['ad_account_id']
    sender     = row['Sender']

    print(f"\n{'='*70}")
    print(f"Query Run for {brand}")
    print(f"{'='*70}\n")

    # ── Pre-check: skip brand if all campaigns for today's action_date are already done ──
    try:
        conn_check = psycopg2.connect(**Voylla_config)
        cur_check = conn_check.cursor()
        cur_check.execute(f"""
            SELECT COUNT(*)
            FROM {brand}."Blinkit_actions_llm"
            WHERE "Brand" = '{brand}'
              AND action_date::TEXT = (
                  SELECT to_char(date(MAX(action_date)) - INTERVAL '0' DAY, 'YYYY-MM-DD')
                  FROM {brand}."Blinkit_actions_llm"
                  WHERE "Brand" = '{brand}'
              )
              AND (update_status IS NULL OR update_status != 'done')
              AND current_cpm != 0
              AND action != 'PAUSE'
        """)
        pending_count = cur_check.fetchone()[0]
        cur_check.close()
        conn_check.close()

        if pending_count == 0:
            print(f"✅ All campaigns already done for {brand} — skipping login.")
            continue
        else:
            print(f"🔄 {pending_count} pending rows for {brand} — proceeding.")
    except Exception as e:
        print(f"⚠️ Pre-check failed for {brand}, proceeding anyway: {e}")

    # ── Selenium setup (once per brand) ───────────────────────────────────────
    options = Options()
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-gpu")
    options.add_argument(
        "user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36"
    )
    options.set_capability("goog:loggingPrefs", {"performance": "ALL"})

    driver = webdriver.Chrome(options=options)
    driver.execute_cdp_cmd("Network.enable", {})

    token = _cfuvid = __cfruid = __cf_bm = None
    SUBJECT_KEYWORD = "Blinkit Brand Central"
    TO_EMAIL = ad_account

    # ── Login retry loop ──────────────────────────────────────────────────────
    LOGIN_ATTEMPTS = 3
    for login_attempt in range(1, LOGIN_ATTEMPTS + 1):
        try:
            driver.get("https://brands.blinkit.com/")
            time.sleep(2)
            driver.get("https://brands.blinkit.com/")
            time.sleep(5)

            wait = WebDriverWait(driver, 20)

            login_btn = WebDriverWait(driver, 10).until(
                EC.element_to_be_clickable((By.XPATH, "//button[.//div[text()='Login']]"))
            )
            login_btn.click()
            time.sleep(2)

            username_field = wait.until(
                EC.visibility_of_element_located((By.CSS_SELECTOR, "input[placeholder='Email']"))
            )
            human_typing_advanced(username_field, ad_account)

            request_link = WebDriverWait(driver, 10).until(
                EC.element_to_be_clickable((By.XPATH, "//span[text()='Request Sign in Link']"))
            )
            request_link.click()
            time.sleep(15)

            import imaplib
            import email
            from bs4 import BeautifulSoup
            from datetime import datetime, timedelta
            from selenium import webdriver

            IMAP_SERVER = "imap.gmail.com"
            SENDER = sender
            SUBJECT_KEYWORD = "Blinkit Brand Central"

            def extract_sign_in_link():
                mail = imaplib.IMAP4_SSL(IMAP_SERVER)
                mail.login(email_1, password_1)
                mail.select("INBOX")
                since_date = (datetime.now() - timedelta(days=1)).strftime("%d-%b-%Y")
                search_criteria = (
                    f'(FROM "{sender}" TO "{TO_EMAIL}" '
                    f'SUBJECT "{SUBJECT_KEYWORD}" SINCE "{since_date}")'
                )
                status, messages = mail.search(None, search_criteria)
                if status != "OK":
                    mail.logout()
                    raise Exception("Email search failed")
                email_ids = messages[0].split()
                if not email_ids:
                    mail.logout()
                    raise Exception("No matching email found")
                status, msg_data = mail.fetch(email_ids[-1], "(RFC822)")
                if status != "OK":
                    mail.logout()
                    raise Exception("Failed to fetch email")
                msg = email.message_from_bytes(msg_data[0][1])
                sign_in_link = None
                for part in msg.walk():
                    if part.get_content_type() == "text/html":
                        html = part.get_payload(decode=True)
                        if not html:
                            continue
                        soup = BeautifulSoup(html.decode(errors="ignore"), "html.parser")
                        for a in soup.find_all("a", href=True):
                            if "sign in to your account" in a.get_text(strip=True).lower():
                                sign_in_link = a["href"]
                                break
                    if sign_in_link:
                        break
                mail.logout()
                if not sign_in_link:
                    raise Exception("Sign-in link not found")
                return sign_in_link

            link = extract_sign_in_link()
            print("✅ Sign-in link extracted")

            driver.get(link)
            time.sleep(2)
            driver.get("https://brands.blinkit.com/diy/list")
            time.sleep(2)

            all_cookies  = driver.get_cookies()
            cookies_dict = {c['name']: c['value'] for c in all_cookies}
            for c_name, c_value in cookies_dict.items():
                if c_name == "_cfuvid":    _cfuvid  = c_value
                elif c_name == "__cfruid": __cfruid = c_value
                elif c_name == "__cf_bm":  __cf_bm  = c_value

            token = extract_firebase_token(driver)

            if token:
                print("✅ Firebase token captured")
                break
            else:
                print(f"❌ Firebase token not found (attempt {login_attempt}/{LOGIN_ATTEMPTS})")
                time.sleep(50)

        except Exception as e:
            print(f"❌ Login attempt {login_attempt} failed: {e}")
            traceback.print_exc()

    if not token:
        print(f"❌ Could not obtain token for {brand} — skipping brand")
        driver.quit()
        continue


    query3 = f"""
    WITH touched_campaigns AS (
    SELECT DISTINCT campaign_id
    FROM {brand}."Blinkit_actions_llm"
    WHERE "Brand" = '{brand}'
      AND user_implemented IN ('true','false')
      AND action_date = (
          SELECT 
			 --MAX(action_date)
          to_char(date(MAX(action_date)) - INTERVAL '0' DAY,'YYYY-MM-DD')::TEXT
          FROM {brand}."Blinkit_actions_llm"
          WHERE "Brand" = '{brand}'
      )
),

latest_actions AS (
    SELECT *
    FROM {brand}."Blinkit_actions_llm"
    WHERE "Brand" = '{brand}'
      AND action_date = (
          SELECT 
			 --MAX(action_date)
          to_char(date(MAX(action_date)) - INTERVAL '0' DAY,'YYYY-MM-DD')::TEXT
          FROM {brand}."Blinkit_actions_llm"
          WHERE "Brand" = '{brand}'
      )
      AND campaign_id IN (SELECT campaign_id FROM touched_campaigns)
),

updated_keywords AS (
    SELECT 
        la.campaign_id,
        REPLACE(targeting,'_',' ') AS targeting,
        CASE 
            WHEN "action" LIKE 'INCREASE%' 
                THEN (bc.cpm::FLOAT * (1 + 10::FLOAT/100))::INT
            WHEN "action" LIKE 'DECREASE%' 
                THEN (bc.cpm::FLOAT * (1 - 10::FLOAT/100))::INT
            ELSE bc.cpm
        END AS updated_cpm,   bc.budget as campaign_budget
           
    FROM latest_actions la
    left join {brand}."Blinkit_CPM" bc ON la."campaign_id" = bc.campaign_id AND REPLACE(targeting,'_',' ') = bc.keyword
   -- WHERE user_implemented = 'true'
)

SELECT 
    la.action_date,
    la.campaign_id,
    la.campaign_name,
    uk.campaign_budget,

    to_char(DATE_ADD(id.start_ts, INTERVAL '1' DAY), 'MM/DD/YYYY') AS start_ts,
    to_char(id.end_ts, 'MM/DD/YYYY') AS end_ts,

    id.campaign_type,
    id.objective_type,
    id.product_id,

    REPLACE(la.targeting,'_',' ') AS targeting,

    la."action",
    la.bid_change,
    la.user_implemented,
    la.override_note,
    la.current_cpm,
    COALESCE(uk.updated_cpm, la.current_cpm) AS "Updated_CPM",
    region_type, case when region_ids = '' THEN '-1' ELSE region_ids END AS "region_ids",
    la."Brand"

FROM latest_actions la

LEFT JOIN updated_keywords uk
    ON la.campaign_id = uk.campaign_id
    AND REPLACE(la.targeting,'_',' ') = uk.targeting

LEFT JOIN (
    SELECT DISTINCT 
        campaign_id, campaign_budget, start_ts, end_ts, 
        campaign_type, objective_type, product_id, region_type,
        TRANSLATE(region_ids::TEXT, '{{}}', '') AS region_ids
    FROM {brand}."Blinkit_CampaignWise_ProductID" i
    WHERE i.region_type IS NOT null
) id 
ON la.campaign_id::TEXT = id.campaign_id::TEXT

WHERE la.current_cpm <> 0
AND la.action <> 'PAUSE'
AND uk.campaign_budget IS NOT null
AND (la.update_status IS NULL OR la.update_status != 'done')
--AND la.campaign_id = '296468'

ORDER BY la.action_date DESC;
    
    """

    import datetime 
    try:
        print("Query Runing Started at: "+datetime.datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S'))
        conn_misauto = psycopg2.connect(**Voylla_config)
        cursor = conn_misauto.cursor()
        cursor.execute(query3)
        print("Query Runing completed at: "+datetime.datetime.fromtimestamp(time.time()).strftime('%Y-%m-%d %H:%M:%S'))
        
        data = cursor.fetchall()
        column = [col[0] for col in cursor.description]
        df_campaign_details = pd.DataFrame(data, columns = column)
        
        if df.empty:
            with yagmail.SMTP(Alert_email, Alert_password) as yag:
                subject=f"{SD['Script_Name']} - Query Raw count: {len(df_campaign_details)}"
                body= f"""Hi Team,
                
                There is no data fetched by using below query:-
                
                Query:- {query3}
                
                """
                yag.send(to=receiver_email,subject=subject, contents=body
                         ,headers={"From": f"{Alert_sender_name} <{Alert_email}>"})
                print(f"Raw count: {len(df_campaign_details)}")
        else:
            print(f"Raw count: {len(df_campaign_details)}")
    
    except psycopg2.Error as error:
        body = f'Error executing SQL query: {str(error)}'
        print(body)
        with yagmail.SMTP(Alert_email, Alert_password) as yag:
            subject=SD['Script_Name']+"- Query Error"
            yag.send(to=receiver_email,cc= None,bcc=None, subject=subject, contents=body,
                    headers={"From": f"{Alert_sender_name} <{Alert_email}>"})
    finally:
        cursor.close()

    
    for campaign_id in df_campaign_details['campaign_id'].unique():
    
        df_filtered = df_campaign_details[df_campaign_details['campaign_id'] == campaign_id]
    
        pid_list = df_filtered['product_id'].astype(int).unique().tolist()
        print(pid_list)
        asset_type = df_filtered['campaign_type'].iloc[0]
        objective_type = df_filtered['objective_type'].iloc[0]
        total_budget = int(df_filtered['campaign_budget'].iloc[0])
        start_ts = df_filtered['start_ts'].iloc[0]
        end_ts = df_filtered['end_ts'].iloc[0]
        campaign_name = df_filtered['campaign_name'].iloc[0]
        region_ids = df_filtered['region_ids'].iloc[0]

        # 🔹 Deduplicate keywords (VERY IMPORTANT)
        df_keywords = (
            df_filtered
            .groupby('targeting', as_index=False)
            .agg({
                'Updated_CPM': 'max'
            })
        )
        
        # 🔹 Build keyword structure
        result = []
        for _, row in df_keywords.iterrows():
            keyword_obj = {
                "keyword": row["targeting"],
                "bids": [
                    {
                        "match_type": "EXACT",
                        "cpm": int(row["Updated_CPM"]),
                        "max_boost": None
                    }
                ]
            }
            result.append(keyword_obj)

        url = "https://brands.blinkit.com/adservice/v3/campaigns"
    
        headers = {
            "accept": "application/json, text/plain, */*",
            "content-type": "application/json;charset=UTF-8",
            "origin": "https://brands.blinkit.com",
            "referer": f"https://brands.blinkit.com/diy/edit/campaign/{campaign_id}",
            "user-agent": "Mozilla/5.0",
            "firebase_user_token": token,
        }
    
        cookies = {
            "__cf_bm": __cf_bm,
            "__cfruid": __cfruid,
            "_cfuvid": _cfuvid
        }
    
        payload = {
            "source_platform": "diy_dashboard_web",
            "requested_by": ad_account,
            "advertiser_id": 1073,
            "brand_name": "GBL Altair Private Limited",
            "campaign_id": campaign_id,
            "objective_type": objective_type,
            "asset_type": asset_type,
            "highlighted_pids": ",".join(map(str, pid_list)),
            "campaign_data": {
                "pids": ",".join(map(str, pid_list)),
                "products": [
                    {"id": int(pid), "is_star_product": False}
                    for pid in pid_list
                ],
                "brand_ids": None,
                "category_ids": None
            },
            "campaign_start": start_ts,
            "campaign_end": end_ts,
            "infinite_campaign": True,
            "cpm": 0,
            "name": campaign_name,
            "bidding_strategy": {
                "total_budget": total_budget,
                "pacing_type": "DAILY"
            },
            "campaign_targeting": {
                "city_ids": str(region_ids),
                "is_extendable": False,
                "negative_keywords": [],
                "keyword_targeting": {"keywords": result}
            },
            "campaign_request_type": "RESTART"
        }
    
        try:
            response = requests.put(url, json=payload, headers=headers, cookies=cookies)
            print(payload)
            print(f"Campaign: {campaign_id}")
            print(response.status_code)
            print(response.text)
    
            # ✅ SUCCESS CASE
            if response.status_code in [200, 201]:
                changes_text = format_changes(df_filtered)
                body = f"""
                Hi Team,
                
                ✅ Campaign Update Successful
                
                Campaign ID: {campaign_id}
                Campaign Name: {campaign_name}
                
                📊 Bid Changes:
                
                {changes_text}
                
                """
                send_alert_email("Campaign Success", body, brand)
                update_campaign_status(brand, campaign_id, 'done', Voylla_config)
    
            # ❌ FAILURE CASE (API responded but failed)
            else:
                body = f"""
                Hi Team,
                
                ❌ Campaign Update Failed
                
                Campaign ID: {campaign_id}
                Campaign Name: {campaign_name}
                
                Status Code: {response.status_code}
                
                Response:
                {response.text}
                    
                    """
                send_alert_email("Campaign Failed", body, brand)
                update_campaign_status(brand, campaign_id, 'failed', Voylla_config)
    
        except Exception as e:
            print(f"❌ Exception for campaign {campaign_id}: {e}")
            body = f"""
            Hi Team,
            
            🚨 Exception occurred while updating campaign
            
            Campaign ID: {campaign_id}
            Campaign Name: {campaign_name}
            
            Error:
            {str(e)}
            
    """
            send_alert_email("Campaign Error", body, brand)
            update_campaign_status(brand, campaign_id, 'failed', Voylla_config)
    
        print("------")

    driver.quit()
```
